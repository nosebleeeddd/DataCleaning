## Capitalize each brandname entry

```python
data["BrandName"].apply(lambda x: x.capitalize())
```

```python
# SET THE CHANGES WITH:
data["BrandName"] = data["BrandName"].apply(lambda x: x.capitalize())
```

```python
data
```

## Renaming Nan columns

```python
# CHECK TYPE BEFORE WE MANIPULATE
type(data["Sizes"][27929])
```

```python
# SOME COLUMNS NEED DIFFERENT OPERATIONS
data["Sizes"] = data["Sizes"].replace("Nan", "NA VALUED")
```

```python
data.tail()
```

## DROP NA VALUED we created

```python
# THIS WILL FILTER OUT THE NA VALUED WE CREATED FOR Nan
# we save changes in variable too
data = data[data["Sizes"] != "NA VALUED"]
```

```python
# we can see NA VALUED are dropped
data.tail()
```

```python
# WE DO THE SAME FOR DISCOUNT
data["Discount"] = data["Discount"].replace("Nan", "No Discount")
```

```python
data.tail()
```
