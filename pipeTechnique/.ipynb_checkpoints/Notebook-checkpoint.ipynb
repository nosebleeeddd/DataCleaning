### .pipe makes our data cleaning code readable,clean,reusable

```python
.pipe()
```

```python
# EXAMPLE DICT DATASET
data = {
    "Name": ["Alice", "BOB", "Charlie", None],
    "Age": ["25", "30", "?", "40"],
    "Salary": ["50000", "60000", "not available", "70000"]
}
```

```python
# PASS OUR DICT AS DATAFRAME
df = pd.DataFrame(data)
```

```python
df
```

```python
# GET RID OFF UPPERCASE IN Name Column
df["Name"] = df["Name"].str.strip().str.title()
```

```python
df
```

```python
# GET RID OF None in Name
df = df[df["Name"].notna()]
```

```python
df
```

```python
# CHANGE AGE ? to NaN 
df["Age"] = pd.to_numeric(df["Age"], errors = "coerce")
```

```python
# DROP NaN
df = df.dropna(subset= ["Age"])
df["Salary"] = pd.to_numeric(df["Salary"], errors= "coerce")
```

```python
df
```

### - code above not scalable/ hard to debug. NOW WE USE PIPE

## pipe lets you pass dataframes thru custom functions

```python
def clean_names(df):
    df = df.copy()
    df["Name"] = df["Name"].str.strip().str.title()
    return df[df["Name"].notna()]

def clean_age(df):
    df = df.copy()
    df["Age"] = pd.to_numeric(df["Age"], errors = "coerce")
    return df.dropna(subset = ["Age"])

def clean_salary(df):
    df = df.copy()
    df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
    return df

```

```python
clean_df = (
    df
    .pipe(clean_names)
    .pipe(clean_age)
    .pipe(clean_salary)
)
```

```python
clean_df
```

```python
df
```

```python
# APPLYING TO UNCLEANED DATAFRAME
df.pipe(clean_names).pipe(clean_age).pipe(clean_salary)
```

```python
# Create another function for min age
def filter_min_age(df, min_age):
    return df[df["Age"] >= min_age]
```

```python
df.pipe(clean_names).pipe(clean_age).pipe(clean_salary).pipe(filter_min_age, min_age = 27)
```
