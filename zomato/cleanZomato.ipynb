# 2ND WAY
# Dropping the columns

```python
columns_to_drop
```

# REMOVING 1 BY 1

```python
data.rename(columns = {"name":"name1","online_order":"onlineorder123"})
```

```python
data.columns
```

```python
# capitalizes starting char in columns
data.columns[0].capitalize()
```

```python
# capitalizes the column names
for i in data.columns:
    print(i.capitalize())
```

```python
# we append capitalized data to a list
new_column_names = []
for i in data.columns:
    new_column_names.append(i.capitalize())
```

```python
new_column_names
```

```python
# set the changes of the column names
data.columns = new_column_names
```

```python
data.columns
```

# GET RID OF DUPLICATES

```python
# Finding duplicates
#data.duplicated()
data.duplicated().sum()
```

```python
# Find duplcates boolean
data.duplicated().value_counts()
data.duplicated().head(15)
```

```python
# DROP THE DUPLICATES
#data.drop_duplicates()                       # creates copy of df no duplicates
data.drop_duplicates(inplace = True)         # inplace applies changes
```

```python
# Check if duplicates were removed
data.duplicated().sum()
```

# REMOVE NaN VALUES

```python
# Gets boolean values of NaN
data.isna()
# Gets sum of NaN values
data.isna().sum()
```

```python
# DROP the NaN values
#data.dropna()                   # creates a copy of df with no NaN values
data.dropna(inplace = True)
```

```python
# Check if NaN values were removed
data.isna().sum()
```

```python
data
```

# FIX THE INDEXES BECAUSE WE REMOVED NaN and Dupes

```python
# RESET INDEX FOR PROPER INDEXING
#data.reset_index()                              # creates copy
data.reset_index(inplace=True)                   # applys changes to df
```

```python
# must drop old index
data.drop(columns= "index", inplace=True)
```

```python
data
```

# CLEANING INDIVIDUAL COLUMNS

```python
# Clean the rate column
#type(data["Rate"][14])
data["Rate"][14]
```

```python
# displays rate series
data["Rate"]
```

```python
# 1. fix spacing for column cleaning
data["Rate"].str.replace(" ", "")
```

# CHECK FOR SOME TRANSFORMATIONS

```python
# Change text columns to numerics so models can read them
# 1 = yes , 0 = no
data["Online_order"]
```

```python
# lambda to change column
data["Online_order"].apply(lambda x: 1 if x == "Yes" else 0)
```

```python
# APPLY TO THE DATAFRAME
data["Online_order"] = data["Online_order"].apply(lambda x: 1 if x == "Yes" else 0)
```
